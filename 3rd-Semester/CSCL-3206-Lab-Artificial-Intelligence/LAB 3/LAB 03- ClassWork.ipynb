Using dict() - Constructors

```python
teacher = dict(
   name="kari",
    age=33,
    subject="AI"
    
)
print(teacher)
```

Using tuple to convert into dict()

```python
 teacher = dict([
  {"name","kari"},
    {"age",33},
    {"subject","AI"}
 ]
)
print(teacher)
```

Using get() to access value

```python
student={
    "Name": "Ali",
    "Age":21,
    "Program":"Bs-AI"
}
print("Age:",student.get("Age"))
```

Retriving keys - values - items - Age

```python
student={
    "Name": "Ali",
    "Age":21,
    "Program":"Bs-AI"
}
print("Keys:",student.keys())
print("Values:",student.values())
print("Items:",student.items())
print("Age:",student.get("Age"))
```

Adding items in dictionary

```python
student['Semseter']="3rd"
print(student)
```

If the keys and values already exist, it will modify its pair.


```python
student["Age"]=44
print(student)
```

Removing any item from the dict using pop() and del

```python
student={
    "Name": "Ali",
    "Age":21,
    "Program":"Bs-AI"
}
removed= student.pop("Age")

print("Removed value: ", removed)
print("Sucessfully deleted", student)
```

Looping using key

```python
student={
    "Name": "Ali",
    "Age":21,
    "Program":"Bs-AI"
}
for a in student:
    print(student[a])
```

Looping using values

```python
Marks={
    "DLD": 33,
    "DSA":21,
    "DS":45
}
for a in Marks.values():
    print([a])
```
